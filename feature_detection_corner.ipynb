```python
import os
import cv2
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10, 10))
folder_path = "images"
images_files = os.listdir(folder_path)

for i, filename in enumerate(images_files):
    img = cv2.imread(os.path.join(folder_path, filename))

    fast = cv2.FastFeatureDetector_create()
    keypoints_fast = fast.detect(img)
    fast_rest = cv2.drawKeypoints(img, keypoints_fast, None, color=(255, 0, 0))

    orb = cv2.ORB_create()
    keypoints_orb = orb.detect(img)
    orb_rest = cv2.drawKeypoints(img, keypoints_orb, None, color=(0, 0, 255))

    fast_rest = cv2.cvtColor(fast_rest, cv2.COLOR_BGR2RGB)
    orb_rest = cv2.cvtColor(orb_rest, cv2.COLOR_BGR2RGB)

    plt.subplot(len(images_files), 2, i*2 + 1)
    plt.imshow(fast_rest)
    plt.title(f"FAST: {filename}")
    plt.axis("off")     

    plt.subplot(len(images_files), 2, i*2 + 2)
    plt.imshow(orb_rest)
    plt.title(f"ORB: {filename}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```
